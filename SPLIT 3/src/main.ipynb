K-means se basa en la distancia euclidiana entre los puntos de datos. Si dos variables están altamente correlacionadas, la distancia entre los puntos en ese espacio multidimensional podría ser distorsionada. Esto puede hacer que el modelo no distinga correctamente los clusters.

```python
X = df[['Fresh', 'Detergents_Paper']].copy()
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
inertia = []
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

# Graficar el método del codo
plt.plot(range(1, 11), inertia)
plt.xlabel('Número de clusters (K)')
plt.ylabel('Inercia')
plt.title('Método del Codo')
plt.show()
```

```python
# Creando el k-Means para los 4 grupos encontrados
kmeans = KMeans(n_clusters = 5, init = 'k-means++', random_state = 42)
# k-means++ acelera la convergencia, respecto de random
y_kmeans = kmeans.fit_predict(X)
```

```python
y_kmeans
```

```python
df['pred']=y_kmeans
df
```

```python
df.groupby(['pred']).count()
```

```python


plt.scatter(X.loc[y_kmeans == 0, 'Fresh'], X.loc[y_kmeans == 0, 'Detergents_Paper'], s = 100, c = 'red', label = 'Cluster 1')
plt.scatter(X.loc[y_kmeans == 1, 'Fresh'], X.loc[y_kmeans == 1, 'Detergents_Paper'], s = 100, c = 'blue', label = 'Cluster 2')
plt.scatter(X.loc[y_kmeans == 2, 'Fresh'], X.loc[y_kmeans == 2, 'Detergents_Paper'], s = 100, c = 'green', label = 'Cluster 3')
plt.scatter(X.loc[y_kmeans == 3, 'Fresh'], X.loc[y_kmeans == 3, 'Detergents_Paper'], s = 100, c = 'cyan', label = 'Cluster 4')
plt.scatter(X.loc[y_kmeans == 4, 'Fresh'], X.loc[y_kmeans == 4, 'Detergents_Paper'], s = 100, c = 'magenta', label = 'Cluster 5')
#plt.scatter(X.loc[y_kmeans == 5, 'bmi'], X.loc[y_kmeans == 5, 'charges'], s = 100, c = 'yellow', label = 'Cluster 6')


plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s = 100, c = 'black', label = 'Centroids') # changed color to black for contrast


plt.title('Clusters of customers')
plt.xlabel('Fresh') # Updated label for clarity
plt.ylabel('Detergents_Paper') # Updated label for clarity
plt.legend()
plt.show()
```

Vamos a probar el siguiente algoritmo para obtener otra perspectiva sobre el aprendizaje no supervisado.  
Uno de los principales problemas de K-Means es que funciona bien cuando los datos tienen una distribución esférica alrededor de los centroides. Sin embargo, es necesario determinar un valor óptimo de *K*.  

Como alternativa, se suele utilizar un algoritmo de clustering basado en densidades.  

El siguiente algoritmo es **DBSCAN**.


## 🏷️ La técnica de DBSCAN

```python
from sklearn.cluster import DBSCAN

# Definir el modelo DBSCAN
#dbscan = DBSCAN(eps=5000, min_samples=5)  # Ajusta 'eps' según los datos 1 SOLO GRUPO!!!
#dbscan = DBSCAN(eps=2000, min_samples=5)  # Ajusta 'eps' según los datos 2 GRUPOS DISPARES!
dbscan = DBSCAN(eps=3500, min_samples=5)  # Ajusta 'eps' según los datos 2 GRUPOS DISPARES!


# Ajustar el modelo
Z_dbscan = dbscan.fit_predict(W)

# Identificar los clusters
clusters = np.unique(Z_dbscan)
```

```python
colors = ['red', 'blue', 'green', 'cyan', 'magenta', 'yellow']
plt.figure(figsize=(8,6))
for i, cluster in enumerate(clusters):
    color = 'black' if cluster == -1 else colors[i % len(colors)]
    label = 'Noise' if cluster == -1 else f'Cluster {cluster}'
    
    plt.scatter(W.loc[Z_dbscan == cluster, 'Grocery'], 
                W.loc[Z_dbscan == cluster, 'Detergents_Paper'], 
                s=100, color=color, label=label)

plt.title('DBSCAN Clustering')
plt.xlabel('Grocery')
plt.ylabel('Detergents_Paper')
plt.legend()
plt.show()
```

```python
from sklearn.neighbors import NearestNeighbors
import numpy as np

# Calcular distancias entre vecinos
neighbors = NearestNeighbors(n_neighbors=5)
neighbors_fit = neighbors.fit(W_scaled)
distances, _ = neighbors_fit.kneighbors(W_scaled)

# Ordenar distancias
distances = np.sort(distances[:, 4], axis=0)

# Graficar la curva de distancias para detectar el "codo"
plt.plot(distances)
plt.xlabel('Puntos ordenados')
plt.ylabel('Distancia al 5to vecino')
plt.title('Método del codo para elegir eps')
plt.show()
```

🤔 **¿Por qué ocurre esto?**  

Hasta un valor de 400, las distancias son bajas, lo que indica la presencia de grupos densos.  

A partir de 400, hay un salto considerable, lo que sugiere que algunos puntos están más aislados y podrían considerarse ruido.  

DBSCAN interpreta estos puntos aislados como ruido en lugar de incluirlos en algún cluster.  

---  

### **Conclusión: ¿K-Means funcionó mejor que DBSCAN en este caso?**  

Los datos presentan una estructura continua, sin separaciones claras en términos de densidad.  

En el gráfico **Grocery vs. Detergents_Paper**, los puntos siguen una distribución lineal en lugar de formar clusters bien definidos con regiones de alta y baja densidad.  

DBSCAN funciona mejor en conjuntos donde los grupos están bien separados y existen zonas vacías entre ellos, lo que no sucede aquí.  

Por otro lado, K-Means asume clusters esféricos y balanceados, lo que se ajusta mejor a la distribución de estos datos.  

- K-Means asigna cada punto a un cluster, incluso si no hay una separación clara.  
- DBSCAN, en cambio, etiqueta muchos puntos como ruido cuando no detecta suficiente densidad.  
- DBSCAN identifica como ruido los puntos aislados debido a la distancia entre ellos.  

El método del codo mostró que las distancias se mantienen bajas hasta aproximadamente **eps ≈ 400**, momento en el que hay un salto brusco.  

Esto indica que algunos puntos están más alejados, lo que lleva a DBSCAN a excluirlos como ruido, mientras que K-Means los asigna a un cluster.  

En datasets con alta variabilidad y sin estructuras densas definidas, K-Means resulta más robusto.  

Dado que en este caso los datos siguen una tendencia general en lugar de formar grupos densos, K-Means logra una partición más útil.  

DBSCAN, al no encontrar áreas claras de alta densidad, genera muy pocos clusters y clasifica una gran cantidad de puntos como ruido.
