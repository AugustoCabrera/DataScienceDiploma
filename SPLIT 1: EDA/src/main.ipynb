Un *"dato no nulo"* significa que la celda en esa posición del DataFrame contiene un valor válido, es decir, no está vacía ni falta. En pandas, cuando una celda tiene un valor faltante o vacío, se considera un "dato nulo" y se representa típicamente como NaN (Not a Number) en el caso de datos numéricos o None para otros tipos de datos.

Entonces, todas las columnas tienen "Non-Null Count" igual a 16,846, lo que indica que **NO** hay datos faltantes o nulos en esas columnas; todas las celdas están llenas con valores válidos.

Un dataset sin valores nulos es de mayor calidad porque facilita el análisis, mejora la precisión de los modelos, y reduce el trabajo adicional necesario para limpiar los datos 🧹

```python
# Supongamos que tu DataFrame se llama df
resultados = {}

# Iterar sobre cada columna, excepto las que no tienen sentido analizar (como 'tot_income', 'gender', etc.)
columnas_a_analizar = [col for col in df.columns if col not in ['tot_income', 'gender', 'weighs', 'year', 'quarter']]

for col in columnas_a_analizar:
    # Agrupar por género y por cada valor único de la columna
    promedio = df.groupby(['gender', col])['tot_income'].mean().reset_index()
    
    # Encontrar el valor con el ingreso promedio más alto para cada género
    max_por_genero = promedio.loc[promedio.groupby('gender')['tot_income'].idxmax()]
    
    # Guardar los resultados en un diccionario
    resultados[col] = max_por_genero[[col, 'gender', 'tot_income']].to_dict(orient='records')

# Mostrar resultados
for col, data in resultados.items():
    print(f"Columna: {col}")
    for entry in data:
        print(f"  Género: {entry['gender']}, Valor: {entry[col]}, Ingreso promedio: {entry['tot_income']}")
```

# 2️⃣ Explorar el set de datos validando características y composición.

Para visualizar la distribución de los ingresos, identificar valores atípicos (outliers), y tener una idea de la mediana, cuartiles y rango intercuartílico de los datos de ingresos. Hacemos:

```python
# Boxplot
plt.figure(figsize=(8, 5))
sns.boxplot(x=df['ingreso_total'], color='lightgreen')
plt.title('Boxplot de Ingresos Totales')
plt.xlabel('Ingreso Total')
plt.show()
```

El boxplot que muestras parece indicar que los datos de ingreso total están altamente concentrados en valores bajos, mientras que hay un número significativo de valores atípicos (outliers) en rangos altos. Esto es común en distribuciones de ingresos, ya que la mayoría de las personas tienen ingresos dentro de un rango más estrecho, y unos pocos casos extremos (como altos ejecutivos o empresarios) tienen ingresos muy elevados.

### Mejora del Análisis del Gráfico

Para entender mejor la distribución:

```python
# Calcular estadísticas descriptivas
Q1 = df['ingreso_total'].quantile(0.25)  # Primer cuartil
Q2 = df['ingreso_total'].median()        # Mediana
Q3 = df['ingreso_total'].quantile(0.75)  # Tercer cuartil
IQR = Q3 - Q1                                 # Rango intercuartílico
upper_bound = Q3 + 1.5 * IQR                  # Límite superior
lower_bound = Q1 - 1.5 * IQR                  # Límite inferior

# Valores superiores al límite superior
outliers = df[df['ingreso_total'] > upper_bound]
num_outliers = len(outliers)

# Mostrar estadísticas
print("Primer cuartil (Q1):", Q1)
print("Mediana (Q2):", Q2)
print("Tercer cuartil (Q3):", Q3)
print("Rango intercuartílico (IQR):", IQR)
print("Límite superior:", upper_bound)
print("Número de valores superiores al límite superior:", num_outliers)

# Crear el boxplot
plt.figure(figsize=(8, 5))
sns.boxplot(x=df['ingreso_total'], color='lightgreen')
plt.title('Boxplot de Ingresos Totales con Límites')
plt.xlabel('Ingreso Total')

# Añadir líneas para los cuartiles y la mediana
plt.axvline(Q1, color='blue', linestyle='--', label='Q1 (25%)')
plt.axvline(Q2, color='red', linestyle='-', label='Mediana (50%)')
plt.axvline(Q3, color='blue', linestyle='--', label='Q3 (75%)')

# Añadir límites al gráfico
plt.axvline(upper_bound, color='orange', linestyle='--', label='Límite Superior')
plt.axvline(lower_bound, color='orange', linestyle='--', label='Límite Inferior')

# Mostrar la leyenda
plt.legend()
plt.show()
```

```python
# Valores dentro de la caja (entre Q1 y Q3)
within_box = df[(df['ingreso_total'] >= Q1) & (df['ingreso_total'] <= Q3)]
num_within_box = len(within_box)

# Valores fuera de los bigotes
outliers = df[(df['ingreso_total'] < lower_bound) | (df['ingreso_total'] > upper_bound)]
num_outliers = len(outliers)

# Total de datos
total_data = len(df)

# Porcentajes
percent_within_box = (num_within_box / total_data) * 100
percent_outliers = (num_outliers / total_data) * 100

# Mostrar resultados
print("Porcentaje de datos dentro de la caja (Q1-Q3): {:.2f}%".format(percent_within_box))
print("Porcentaje de datos fuera de los bigotes: {:.2f}%".format(percent_outliers))
```

```python
# Definir el salario mínimo
salario_minimo = 156000

# Filtrar los datos
superan_salario_minimo = df[df['ingreso_total'] > salario_minimo]
no_superan_salario_minimo = df[df['ingreso_total'] <= salario_minimo]

# Contar los datos
num_superan = len(superan_salario_minimo)
num_no_superan = len(no_superan_salario_minimo)

# Total de datos
total_datos = len(df)

# Calcular porcentajes
percent_superan = (num_superan / total_datos) * 100
percent_no_superan = (num_no_superan / total_datos) * 100

# Mostrar resultados
print("Número de ingresos que superan el salario mínimo:", num_superan)
print("Porcentaje de ingresos que superan el salario mínimo: {:.2f}%".format(percent_superan))
print("Número de ingresos que no superan el salario mínimo:", num_no_superan)
print("Porcentaje de ingresos que no superan el salario mínimo: {:.2f}%".format(percent_no_superan))
```

```python
# Seleccionar solo las columnas relevantes
columnas_relevantes = [
    'género', 'ingreso_total', 'clase_ocupacional', 'nivel_educativo',
    'tipo_empleador', 'asistencia_educativa', 'tipo_institucion_educativa'
]
df_INTERES = df[columnas_relevantes]

# Filtrar filas con ingresos válidos (por ejemplo, mayores a cero)
df_INTERES = df_INTERES[df_INTERES['ingreso_total'] > 0]

# Guardar el nuevo dataset para análisis
df_INTERES.to_csv("DatasetINTERES.csv", index=False)

# Ver las columnas del dataset
print(df_INTERES.columns)
```

La mayoría de los trabajadores tienen un nivel educativo de secundaria o superior 📚, son empleados en el sector privado y asistieron a instituciones públicas. Hay una gran disparidad en los ingresos, lo que refleja una alta desigualdad salarial.

```python
# Crear el diccionario de mapeo con las definiciones
mapeo = {
    'género': {1: 'Masculino', 2: 'Femenino'},
    'clase_ocupacional': {1: 'Empleador', 2: 'Autónomo', 3: 'Trabajador o empleado', 4: 'Trabajador familiar no remunerado', 9: 'No especificado / Sin respuesta'},
    'nivel_educativo': {1: 'Educación Primaria', 2: 'Educación Secundaria', 3: 'Educación Terciaria', 4: 'Título Universitario', 5: 'Título de Posgrado'},
    'asistencia_educativa': {1: 'Sí, actualmente asistiendo', 2: 'No, pero asistió en el pasado', 3: 'Nunca asistió'},
    'tipo_institucion_educativa': {1: 'Pública', 2: 'Privada', 0: 'No especificado / Sin respuesta'},
    'tipo_empleador': {1: 'Público', 2: 'Privado', 3: 'Otro (especificar)'}
}

# Aplicar el mapeo a cada columna relevante en el DataFrame
for columna, diccionario in mapeo.items():
    df_INTERES[columna] = df_INTERES[columna].map(diccionario)

# Verificar el cambio
print(df_INTERES.head())
```

```python
print(df_INTERES['tipo_institucion_educativa'].unique())
```

```python

# Filtrar datos por género
hombres = df_INTERES[df_INTERES['género'] == 'Masculino']
mujeres = df_INTERES[df_INTERES['género'] == 'Femenino']

# Función para analizar cada columna
def analizar_columna(columna, hombres, mujeres):
    # Calcular el ingreso total promedio por cada valor de la columna
    ingreso_promedio_hombres = hombres.groupby(columna)['ingreso_total'].mean()
    ingreso_promedio_mujeres = mujeres.groupby(columna)['ingreso_total'].mean()

    # Crear tablas de frecuencia en porcentaje
    frec_hombres = hombres[columna].value_counts(normalize=True) * 100
    frec_mujeres = mujeres[columna].value_counts(normalize=True) * 100
    
    # Combinar los promedios de ingreso y las frecuencias en un DataFrame
    comparacion_ingreso = pd.DataFrame({
        'Hombres (Promedio Ingreso)': ingreso_promedio_hombres,
        'Mujeres (Promedio Ingreso)': ingreso_promedio_mujeres
    }).fillna(0)

    comparacion_frec = pd.DataFrame({
        'Hombres (%)': frec_hombres,
        'Mujeres (%)': frec_mujeres
    }).fillna(0)

    # Resetear índice para graficar
    comparacion_ingreso = comparacion_ingreso.reset_index().rename(columns={'index': columna})
    comparacion_frec = comparacion_frec.reset_index().rename(columns={'index': columna})

    # Definir colores para hombres y mujeres
    color_hombres = 'rgb(0, 123, 255)'  # Azul
    color_mujeres = 'rgb(255, 99, 132)'  # Rojo

    # Crear subgráfico
    fig = make_subplots(
        rows=1, cols=2,  # Un gráfico con dos columnas
        column_widths=[0.5, 0.5],  # Ajustar el ancho de las columnas
        subplot_titles=[f'Promedio de Ingreso Total por Género en {columna}', f'Distribución de {columna} por Género']
    )

    # Gráfico de Promedio de Ingreso
    fig.add_trace(
        go.Bar(x=comparacion_ingreso[columna], 
               y=comparacion_ingreso['Hombres (Promedio Ingreso)'], 
               name='Hombres (Promedio Ingreso)',
               marker_color=color_hombres,
               text=comparacion_ingreso['Hombres (Promedio Ingreso)'].round(2),
               textposition='auto'),
        row=1, col=1
    )
    fig.add_trace(
        go.Bar(x=comparacion_ingreso[columna], 
               y=comparacion_ingreso['Mujeres (Promedio Ingreso)'], 
               name='Mujeres (Promedio Ingreso)',
               marker_color=color_mujeres,
               text=comparacion_ingreso['Mujeres (Promedio Ingreso)'].round(2),
               textposition='auto'),
        row=1, col=1
    )

    # Gráfico de Distribución por Género
    fig.add_trace(
        go.Bar(x=comparacion_frec[columna], 
               y=comparacion_frec['Hombres (%)'], 
               name='Hombres (%)',
               marker_color=color_hombres,
               text=comparacion_frec['Hombres (%)'].round(2),
               textposition='auto'),
        row=1, col=2
    )
    fig.add_trace(
        go.Bar(x=comparacion_frec[columna], 
               y=comparacion_frec['Mujeres (%)'], 
               name='Mujeres (%)',
               marker_color=color_mujeres,
               text=comparacion_frec['Mujeres (%)'].round(2),
               textposition='auto'),
        row=1, col=2
    )

    # Ajustes de diseño
    fig.update_layout(
        height=500,
        width=1600,
        title_text=f"Análisis de {columna}",
        xaxis_tickangle=45,
        showlegend=True
    )
    
    # Mostrar gráfico
    fig.show()

# Columnas a analizar
columnas = ['clase_ocupacional', 'nivel_educativo', 'tipo_empleador', 
            'asistencia_educativa', 'tipo_institucion_educativa']

# Analizar cada columna
for columna in columnas:
    analizar_columna(columna, hombres, mujeres)
```

```python
# Calcular el rango intercuartílico (IQR) para identificar y eliminar outliers
def eliminar_outliers(df, columna, grupo):
    # Calcular los límites por grupo (en este caso, por género)
    Q1 = df.groupby(grupo)[columna].quantile(0.25)  # Primer cuartil
    Q3 = df.groupby(grupo)[columna].quantile(0.75)  # Tercer cuartil
    IQR = Q3 - Q1  # Rango intercuartílico

    # Límites para los outliers
    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR

    # Crear un filtro para eliminar outliers
    filtro = df.apply(
        lambda x: (x[columna] >= limite_inferior[x[grupo]]) and (x[columna] <= limite_superior[x[grupo]]),
        axis=1
    )

    # Filtrar los datos y devolver un nuevo DataFrame
    return df[filtro]

# Eliminar outliers del DataFrame original
df_SIN_ATIPICOS = eliminar_outliers(df_INTERES, 'ingreso_total', 'género')

# Guardar el nuevo dataset sin datos atípicos
df_SIN_ATIPICOS.to_csv("DatasetSIN_ATIPICOS.csv", index=False)

# Ver las columnas del dataset final
print(df_SIN_ATIPICOS.columns)
```
